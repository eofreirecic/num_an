```python
def fixed_point(iteration_function, x, epsilon):
    iterations = 0
    while True:
        current_iteration_x = iteration_function(x)
        iterations += 1

        if abs(current_iteration_x - x) < epsilon:
            break

        x = current_iteration_x

    return current_iteration_x, iterations
```
