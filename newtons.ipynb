```python
import math
```

```python
def newton(function, derivative, xn, epsilon):

    iterations = 0

    while True:

        x_n_plus_1 = xn - function(xn) / derivative(xn)
        iterations += 1
        print(x_n_plus_1, xn)
        if abs(x_n_plus_1 - xn) < epsilon:
            break

        xn = x_n_plus_1

    return x_n_plus_1, iterations
```

```python
def f(x):
    return x - 4 - math.log(x)
```

```python
def derivate(x):
    return 1 - 1/x
```

```python
epsilon = 0.00001
result = newton(f, derivate, 2, epsilon)
```
