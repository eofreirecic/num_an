# Método da bissecção

```python
import math
```

```python
def same_signal(function, a, b):
    both_positive = function(a) > 0 and function(b) > 0 
    both_negative = function(a) < 0 and function(b) < 0 
    if both_positive or both_negative:
        return True
    return False
```

```python
def bisection(function, a, b, epsilon):

    if same_signal(function, a, b):
        raise Exception("The scalars a and b do not bound a root")

    midpoint = (a + b) / 2
    iterations = 0

    while abs(function(midpoint)) > epsilon:

        if same_signal(function, a, midpoint):
            a = midpoint
        else:
            b = midpoint

        midpoint = (a + b) / 2
        iterations += 1

    return [midpoint, iterations]
```

A função é dada por $f(x) = e^x - x^2 - 3x$.

```python
def f(x):
    return math.exp(x) - x ** 2 - 3 * x
```

```python
result, iterations = bisection(f, -4, -3, 0.00001)
print(result, iterations)
```
